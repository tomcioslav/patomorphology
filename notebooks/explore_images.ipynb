```python
from config import paths
from pato.dataset import DatasetViewer
from pato.utils import split_image
from pato.visualize import show_image
```

```python
dataset = DatasetViewer(root=paths.data_processed / "nmsc-5x")
print(f"Found {len(dataset)} samples; splits: { {k: len(v) for k, v in dataset.metadata.splits.items()} }")
dataset.sample_ids[:5]
```

```python
sample = dataset[0]
show_image(sample.image)
```

```python
show_image(sample.mask)
```

```python
tiles = split_image(sample, target_size=512, overlap=64)
print(f"Split into {len(tiles)} tiles of {tiles[0].image.shape[:2]}")
show_image(tiles[0].image)
```

```python
show_image(tiles[0].mask)
```
