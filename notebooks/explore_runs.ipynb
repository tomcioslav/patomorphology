# Explore runs

Pick any saved training run, load its predictor, run inference on a sample image, and compare against ground truth.

The predictor knows everything pipeline-specific (tile size, normalization, sliding-window inference, etc.) — same call site for any pipeline.

```python
%load_ext autoreload
%autoreload 2
```

```python
from config import paths
from pato.experiments import list_runs, load_run, load_predictor
from pato.source import NMSCDataset
from pato.visualize import show_image
```

## 1. See what's available

```python
runs = list_runs(paths.runs)
print(f"Found {len(runs)} run(s):")
runs
```

## 2. Pick a run and inspect its config

```python
run_name = runs[-1]  # most recent (sorted lexicographically — timestamps sort right)
run_path = paths.runs / run_name
run = load_run(run_path)
print(f"Run: {run.name}")
print(f"Best ckpt: {run.best_checkpoint().name if run.best_checkpoint() else '(none)'}")
run.config
```

## 3. Load the predictor (dispatch by `pipeline` field)

```python
predictor = load_predictor(run_path)
type(predictor).__name__
```

## 4. Pull a sample and predict

`predictor.predict()` accepts a `PatoImage`, a `numpy` array, or a path on disk.

```python
src = NMSCDataset(root=paths.nmsc_5x)
sample = src[0]
predicted_mask = predictor.predict(sample)
print(f"image: {sample.image.shape}, gt mask: {sample.mask.shape}, predicted: {predicted_mask.shape} {predicted_mask.dtype}")
```

## 5. Visualize: source image / ground truth / prediction

```python
show_image(sample.image)
```

```python
show_image(sample.mask)
```

```python
show_image(predicted_mask)
```
